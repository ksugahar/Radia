# Universal Relaxation Network (URN) -- verified results showcase

URN is a physics-informed, attention-based network that fits broadband
frequency-domain impedance / relaxation spectra and synthesizes an equivalent
SPICE circuit. This is a **conservative showcase** of the committed, verified
paper figures; the full corpus (the IEEE-Access paper IN REVISION with its
reviewer-response tracker `paper/REVIEW.md`, the training scripts, the
runtime-import-locked core `universal_relaxation_network.py` +
`relaxation_basis_library.py` that the shipped `urn_fit_from_csv` MCP tool
imports) is **kept** at `examples/universal_relaxation_network/`.

The figures below are the committed corpus results (the training scripts use
torch and are not re-run here -- the paper is in revision, so its results are not
re-rolled). Verified headline: URN improves the broadband impedance fit by
~22.8% on average vs vector fitting (VF) across the 5 real-world datasets, with
the attention/adaptive-basis mechanism the dominant contributor (ablation).

Promoted (showcase) 2026-06-27.

## Reproducing / using URN

The training + SPICE-synthesis scripts are in the kept corpus:
`generate_paper_figures.py` (the 4 figures above), `validate_all_datasets.py` (the URN-vs-VF table), `ablation_study.py`, and `demo_spice_timedomain.py` (URN fit -> `generate_spice_netlist` -> time-domain step/PWM). They need torch and run against the committed `data/real_world/**` CSVs. The shipped MCP tool `urn_fit_from_csv` (radia-mcp) runtime-imports the corpus core.
