This notebook is result-saved. It reads the synchronized JSON archive, shows the migration lanes, and asserts that every current `examples/induction_heating/*.py` file is covered before any deletion is attempted.


```python
from pathlib import Path
import json

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / 'induction_heating_examples_catalog_results.json').exists():
    NOTEBOOK_DIR = Path('docs/induction_heating').resolve()
archive = json.loads((NOTEBOOK_DIR / 'induction_heating_examples_catalog_results.json').read_text(encoding='utf-8'))
print(json.dumps(archive['summary'], ensure_ascii=False, indent=2))
```
