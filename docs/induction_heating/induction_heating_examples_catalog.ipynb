# Induction Heating Migration Catalog

This result-saved notebook renders the source/hash catalog for induction-heating examples and already-migrated validation records. It is synchronized with `induction_heating_examples_catalog_results.json`.

```python
from pathlib import Path
import json
from collections import Counter
from IPython.display import Markdown, display

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / 'induction_heating_examples_catalog_results.json').exists():
    NOTEBOOK_DIR = Path('docs/induction_heating').resolve()
archive = json.loads((NOTEBOOK_DIR / 'induction_heating_examples_catalog_results.json').read_text(encoding='utf-8'))
print(json.dumps(archive['summary'], ensure_ascii=False, indent=2))
```

## Migration Lanes

```python
records = archive['records']
lane_counts = Counter(r['lane'] for r in records)
subdir_counts = Counter(r.get('subdir', '.') for r in records)
assert sum(lane_counts.values()) == archive['summary']['script_count']

def table(headers, rows):
    return '\n'.join(['| ' + ' | '.join(headers) + ' |', '|' + '|'.join(['---'] * len(headers)) + '|'] + ['| ' + ' | '.join(str(c) for c in row) + ' |' for row in rows])

display(Markdown(table(['lane', 'count'], sorted(lane_counts.items()))))
display(Markdown(table(['subdir', 'python_files'], sorted(subdir_counts.items()))))
```

## Python Inventory

```python
rows = []
for r in records:
    src = r.get('source', {})
    artifacts = r.get('sibling_artifacts', r.get('artifacts', []))
    summary = src.get('summary', '').replace('|', '/')[:120]
    rows.append([r['path'], r['lane'], r.get('protected_reference_count', 0), len(artifacts), src.get('line_count', ''), src.get('sha256', '')[:12], summary])
display(Markdown(table(['path', 'lane', 'refs', 'artifacts', 'lines', 'sha256', 'summary'], rows)))
```

## Protected References

```python
rows = []
for r in records:
    if not r.get('protected_reference_count'):
        continue
    first = r.get('protected_references', [{}])[0]
    where = f"{first.get('path', '')}:{first.get('line', '')}" if first else '-'
    rows.append([r['path'], r['protected_reference_count'], where, r['lane']])
display(Markdown(table(['script', 'refs', 'first_ref', 'lane'], rows)))
```

```python
current_paths = {r['path'] for r in records}
assert archive['summary']['script_count'] == len(current_paths)
assert all(r['source']['text'] for r in records)
assert archive['generated_at_utc']
print(f"catalog records: {len(records)} scripts; source lines: {sum(r.get('line_count', 0) for r in records)}; artifacts: {archive['summary'].get('artifact_count')}")
print('deletion gate: keep examples paths until each protected reference is migrated to docs/src/validation_test/panels')
```
