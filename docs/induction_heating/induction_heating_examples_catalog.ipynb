# Induction Heating Promotion Catalog

This result-bearing catalog records the final docs/src/validation split for the closed induction-heating examples batch.

```python
from pathlib import Path
import json
from IPython.display import Markdown, display

data = json.loads(Path('induction_heating_examples_catalog_results.json').read_text(encoding='utf-8'))
data['summary']
```

```python
def table(headers, rows):
    lines = ['| ' + ' | '.join(headers) + ' |', '|' + '|'.join(['---'] * len(headers)) + '|']
    lines += ['| ' + ' | '.join(str(c) for c in row) + ' |' for row in rows]
    return '\n'.join(lines)
rows = [[a['kind'], a['path'], a['role'], a['exists'], a['sha256'][:12]] for a in data['artifacts']]
display(Markdown(table(['kind', 'path', 'role', 'exists', 'sha256'], rows)))
```

```python
assert data['status'] == 'closed_no_live_induction_heating_examples_topic'
assert data['checks']['all_artifacts_exist']
assert data['checks']['examples_target_count'] == 0
print('catalog checks: PASS')
```
