# Kelvin AdaptiveMesh archive

This notebook promotes the `examples/kelvin_transformation/AdaptiveMesh` Python corpus into a docs-layer archive before pruning repetitive scripts. It stores full source text and SHA-256 hashes in `kelvin_adaptive_mesh_archive_results.json` so the tracked tree does not need to keep every per-order generated runner forever.

The first deletion batch is intentionally narrow: repetitive scripts under `AdaptiveMesh/**/order=*/...py`. Higher-level drivers such as `compare_convergence.py`, `CubeMesh.py`, and TEAM7 modules remain until they are separately lifted into `src`, `validation_test`, or a focused notebook.

```python
from pathlib import Path
import json
import sys

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / 'kelvin_examples_migration.py').exists():
    NOTEBOOK_DIR = Path('docs/kelvin').resolve()
sys.path.insert(0, str(NOTEBOOK_DIR))

from kelvin_examples_migration import build_git_source_archive, markdown_table, write_report_json

archive_path = NOTEBOOK_DIR / 'kelvin_adaptive_mesh_archive_results.json'
fresh_archive = build_git_source_archive(
    prefix='examples/kelvin_transformation/AdaptiveMesh',
    lane='collapse_to_notebook_or_memory',
    include_source=True,
)
archive = fresh_archive
if archive_path.exists():
    existing = json.loads(archive_path.read_text(encoding='utf-8'))
    if existing.get('summary', {}).get('archived_files', 0) > fresh_archive['summary']['archived_files']:
        archive = existing
        print('loaded existing archive to preserve deleted source snapshots')
write_report_json(archive, archive_path)
print(json.dumps(archive['summary'], ensure_ascii=False, indent=2))
print(f'\nwrote {archive_path.relative_to(NOTEBOOK_DIR)}')
```

```python
from IPython.display import Markdown, display

display(Markdown('## Archived source groups'))
display(Markdown(markdown_table([
    {'group': k, 'files': v} for k, v in archive['summary']['by_group'].items()
], ['group', 'files'], max_rows=20)))

display(Markdown('## First safe deletion batch'))
delete_batch = [
    row for row in archive['files']
    if '/order=' in row['path'].replace('\\\\', '/')
]
display(Markdown(markdown_table([
    {
        'path': row['path'],
        'lines': row['source']['line_count'],
        'sha256': row['source']['sha256'][:12],
    }
    for row in delete_batch
], ['path', 'lines', 'sha256'], max_rows=80)))
print(f'first deletion batch size: {len(delete_batch)}')
```

## Deletion rule

After this notebook and JSON are committed, the `order=*` runner scripts can be removed if no external references appear. Keep the aggregate comparison scripts and TEAM7 modules until they are separately represented as notebooks or validation tests.
