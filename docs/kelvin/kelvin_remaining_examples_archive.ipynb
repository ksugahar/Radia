# Kelvin Remaining Examples Archive

This result-bearing notebook is the final migration ledger for the remaining `examples/kelvin_transformation/**/*.py` scripts. It pairs Markdown policy notes with an executable inventory loaded from `kelvin_remaining_examples_archive_results.json`.

The migration rule is intentionally strict: source code is preserved here with SHA-256 before standalone example scripts are removed. Scripts that are actual regression gates move to `validation_test/`; exploratory or plot-only scripts become docs-layer source records rather than live examples.


## Promotion Decision

| Lane | Treatment |
|---|---|
| `validation_test_relocated` | The script remains executable, but under `validation_test/cubit/kelvin_1_4_p_convergence/`; pytest points to the promoted copy. |
| `validation_candidate_archived` | The historical validation probe is archived with full source. A future maintained test should be written against `src/radia` APIs rather than keeping a research script in `examples`. |
| `adaptive_mesh_research_archive` / `adaptive_mesh_plot_archive` | The adaptive-mesh derivation or plotting script is archived as a research record. The lesson belongs in docs; reusable algorithms should be promoted separately to `src`. |
| `team7_research_archive` | TEAM 7 exploratory scripts are archived together because they import each other and are not a shipped panel or fast regression gate. |

This notebook does not rerun the heavy FEM/Cubit/TEAM7 workflows. Its executed result is the source inventory, hashes, and migration table; numerical regression behavior is guarded by the promoted validation tests where appropriate.


```python
from pathlib import Path
import json

candidates = [
    Path('kelvin_remaining_examples_archive_results.json'),
    Path('docs/kelvin/kelvin_remaining_examples_archive_results.json'),
]
archive = next((p for p in candidates if p.exists()), None)
assert archive is not None, f'archive JSON not found in {candidates}'
data = json.loads(archive.read_text(encoding='utf-8'))
summary = data['summary']
print('archive:', archive)
print('record_count:', summary['record_count'])
print('validation_test_relocated_count:', summary['validation_test_relocated_count'])
print('archived_only_count:', summary['archived_only_count'])
print('total_source_lines:', summary['total_source_lines'])
print('total_source_bytes_utf8:', summary['total_source_bytes_utf8'])
assert summary['record_count'] == 22
assert all(r['source'] and len(r['sha256']) == 64 for r in data['records'])
```

```python
from collections import Counter
from IPython.display import Markdown, display

lane_counts = Counter(r['promotion_lane'] for r in data['records'])
rows = ['| Promotion lane | Count |', '|---|---:|']
for lane, count in sorted(lane_counts.items()):
    rows.append(f'| `{lane}` | {count} |')
display(Markdown('\n'.join(rows)))

sample_rows = ['| Path | Lane | Lines | SHA-256 prefix | Promoted to |', '|---|---|---:|---|---|']
for r in data['records']:
    promoted = r.get('promoted_to') or ''
    sample_rows.append(f"| `{r['path']}` | `{r['promotion_lane']}` | {r['line_count']} | `{r['sha256'][:12]}` | `{promoted}` |")
display(Markdown('\n'.join(sample_rows)))
```
