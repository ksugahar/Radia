# Hysteresis models: Play & Energy validation gallery

Radia ships two B-input hysteresis material models -- the **Play model** (`rad.MatPlayHysteresis`, O(K) direct forward `B -> H`) and the **Energy model** (`rad.MatEnergyHysteresis`, Egger Schur-complement Newton) -- plus the **BQM** vector reference and full state management (`MatHysSaveState` / `RestoreState` / `CommitState`, `MatMvsH`). This notebook collects the committed validation figures that lock these models against each other and against measured B-H data.

*Validation-class corpus kept at `validation_test/hysteresis/` (the C++ kernel goldens import `binput_play_fixture.npz`; figure-regenerating scripts need the measured B-H source data). This notebook is the rendered evidence gallery.*
