# VIM Examples Retirement

This result-saved notebook records the retirement of `examples/vim`. The implementation now lives in `src/radia/vim`, `src/radia/mmmm2d.py`, C++ core kernels, and validation goldens under `validation_test/feec`. Old prototype files are intentionally not kept as a source-tree archive; git history is the recovery path.

```python
from pathlib import Path
import json
from IPython.display import Markdown, display

data = json.loads(Path('vim_examples_retirement_results.json').read_text(encoding='utf-8'))
data['summary']
```

```python
def table(headers, rows):
    lines = ['| ' + ' | '.join(headers) + ' |', '|' + '|'.join(['---'] * len(headers)) + '|']
    lines += ['| ' + ' | '.join(str(c) for c in row) + ' |' for row in rows]
    return '\n'.join(lines)

rows = [
    ['validation legacy', data['summary']['legacy_validation_files'], data['summary']['legacy_validation_py']],
    ['delete from examples', data['summary']['deleted_examples_files'], data['summary']['deleted_examples_py']],
]
display(Markdown(table(['lane', 'files', 'py'], rows)))
```

```python
legacy = [f for f in data['legacy_validation_files'] if f['suffix'] == '.py']
rows = [[Path(f['path']).name, f['bytes'], f['sha256'][:12]] for f in legacy]
display(Markdown(table(['legacy module', 'bytes', 'sha256'], rows)))
```

```python
deleted_by_suffix = {}
for f in data['deleted_from_examples']:
    deleted_by_suffix[f['suffix']] = deleted_by_suffix.get(f['suffix'], 0) + 1
assert data['policy']['old_implementation_retention'] == 'git_history_only'
assert data['summary']['legacy_validation_py'] == 18
assert data['summary']['deleted_examples_py'] > 0
print('vim examples retirement ledger checks: PASS', deleted_by_suffix)
```
