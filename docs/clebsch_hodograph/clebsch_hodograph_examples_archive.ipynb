# Clebsch Hodograph Examples Archive

This notebook is a result-saved promotion archive for `examples/clebsch_hodograph`.
It keeps theory, source inventory, result payloads, and placement policy in one
human-readable surface while the adjacent JSON remains the durable debug record.

## Theory and Scope

- Clebsch and hodograph coordinates provide an inverse-design language for magnetic pole shapes: field-line coordinates, scalar potentials, and geometry are tied by pullback/Hodge weights.
- The corpus spans A-method duality, Kelvin/open-boundary variants, accelerator pole design, 3D end effects, saturation loops, and the HDiv-VIM Clebsch bridge.
- This archive is the docs-first layer: full source and historical JSON/media hashes are preserved before any later pruning decision.

## Placement Policy

KEEP/protected for now: validation_test/feec/test_clebsch_hodograph_research.py and multiple docs/MCP knowledge files cite this corpus. Prune only after named references are migrated to docs, src, or validation_test.

## Synchronization

The durable archive is `clebsch_hodograph_examples_archive_results.json`. It stores full source, SHA-256 hashes,
existing result JSON payloads, media hashes, runtime metadata, and the promotion
lane for every script in this batch.


```python
from pathlib import Path
import json
from IPython.display import Markdown, display

archive_path = Path('clebsch_hodograph_examples_archive_results.json')
if not archive_path.exists():
    archive_path = Path('docs/clebsch_hodograph/clebsch_hodograph_examples_archive_results.json')
data = json.loads(archive_path.read_text(encoding='utf-8'))
counts = data['counts']
print(f"topic={data['topic']}")
print(f"python_scripts={counts['python_scripts']}")
print(f"topic_json_files={counts['topic_json_files']}")
print(f"media_files={counts['media_files']}")
print(f"promotion_lane={data['promotion_lane']}")
```

## Script Inventory
Generated from the synchronized JSON archive.

```python
header = '| # | script | lines | json refs | docstring |\n|---:|---|---:|---:|---|'
body = []
for i, rec in enumerate(data['records'], 1):
    doc = (rec.get('docstring_first_line') or '').replace('|', '\\|')
    body.append(f"| {i} | `{rec['path']}` | {rec['line_count']} | {len(rec.get('attached_json', []))} | {doc} |")
display(Markdown('\n'.join([header] + body)))
```

## Integrity Checks

```python
assert counts['python_scripts'] == len(data['records'])
assert all(len(rec['sha256']) == 64 for rec in data['records'])
assert all(rec.get('source') for rec in data['records'])
print('archive integrity: ok')
```

## Result Payloads and Media

```python
rows = data.get('topic_json_files', [])
if rows:
    header = '| # | json | sha256 | keys |\n|---:|---|---|---|'
    body = []
    for i, row in enumerate(rows, 1):
        payload = row.get('payload')
        keys = ', '.join(list(payload.keys())[:8]) if isinstance(payload, dict) else type(payload).__name__
        body.append(f"| {i} | `{row['path']}` | `{row['sha256'][:12]}` | {keys.replace('|', '\\|')} |")
    display(Markdown('\n'.join([header] + body)))
else:
    print('no existing JSON payloads found')

media = data.get('media_files', [])
print(f"media artifacts indexed: {len(media)}")
```
