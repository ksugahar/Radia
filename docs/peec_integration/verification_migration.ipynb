# PEEC verification migration

This result-bearing notebook records the migration of the PEEC verification corpus out of `examples/`. The executable checks now live under `validation_test/peec_integration/verification`, while reusable centerline-mesh reading is promoted to `radia.peec_mesh_import`. The adjacent JSON keeps the full pre-migration source text, hashes, media hashes, routing lane, and post-migration checks.

```python
from pathlib import Path
import json, platform, sys
repo = Path.cwd()
result_path = repo / 'docs/peec_integration/verification_migration_results.json'
data = json.loads(result_path.read_text(encoding='utf-8'))
print(json.dumps({
    'generated_at_utc': data['generated_at_utc'],
    'checked_at_utc': data['post_migration']['checked_at_utc'],
    'python': sys.version.split()[0],
    'platform': platform.platform(),
    'radia_version': data['runtime']['radia_version'],
    'git_head': data['runtime']['git_head'],
    'git_dirty': data['runtime']['git_dirty'],
}, indent=2, ensure_ascii=False))
```

## Routing summary

Validation-class executables and their Cubit/GMSH assets move to `validation_test`. `gmsh_centerline_reader.py` is not kept as a notebook-side helper because it is a reusable mesh import API. `check_funcs.py` is a distilled delete: its purpose was one-off Radia symbol introspection, already recorded in `memory/peec_integration_cleanup_routing.md`.

```python
from IPython.display import Markdown, display
summary = data['summary']
lines = ['| Item | Value |', '| --- | ---: |']
lines.append(f"| Archived tracked files | {summary['total_files']} |")
lines.append(f"| Python files | {summary['python_files']} |")
lines.append(f"| Media/mesh assets | {summary['media_or_mesh_assets']} |")
for lane, count in sorted(summary['by_lane'].items()):
    lines.append(f'| Lane: `{lane}` | {count} |')
display(Markdown('\n'.join(lines)))
```

```python
rows = []
for f in data['files']:
    rows.append((f['lane'], f['old_path'], f['target_path'], f['reason']))
rows = sorted(rows)
lines = ['| Lane | Old path | Target | Reason |', '| --- | --- | --- | --- |']
for lane, old, target, reason in rows[:20]:
    lines.append(f'| `{lane}` | `{old}` | `{target}` | {reason} |')
display(Markdown('\n'.join(lines)))
print(f'showing 20 of {len(rows)} archived files')
```

```python
post = data['post_migration']
assert not post['old_source_root_exists'], post
assert post['validation_root_exists'], post
assert post['validation_file_count'] == data['summary']['by_lane']['validation_test'], post
assert not post['missing_validation_targets'], post['missing_validation_targets']
assert not post['old_paths_still_present'], post['old_paths_still_present']
assert post['src_api_class_present'] and post['src_api_helper_present'], post
assert post['distill_delete_removed'], post
print('migration checks: PASS')
print(json.dumps({
    'validation_file_count': post['validation_file_count'],
    'old_reference_count_after_migration': post['old_reference_count_after_migration'],
    'src_api': 'src/radia/peec_mesh_import.py',
}, indent=2, ensure_ascii=False))
```

```python
media = data['media_hashes']
lines = ['| Asset | Target | SHA-256 prefix | Bytes |', '| --- | --- | ---: | ---: |']
for item in media:
    lines.append(f"| `{item['path']}` | `{item['target_path']}` | `{item['sha256'][:12]}` | {item['size_bytes']} |")
display(Markdown('\n'.join(lines)))
```
