# PEEC Integration Examples Catalog

Result-saved catalog for examples/peec_integration. It stores the full source text, hashes, artifacts, protected references, and migration lanes with current migration lanes.

```python
from pathlib import Path
import json
from IPython.display import Markdown, display
NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / 'examples_catalog_results.json').exists():
    NOTEBOOK_DIR = Path('docs/peec_integration').resolve()
archive = json.loads((NOTEBOOK_DIR / 'examples_catalog_results.json').read_text(encoding='utf-8'))
print(json.dumps(archive['summary'], indent=2, ensure_ascii=False))
```

```python
rows = sorted(archive['summary']['lane_counts'].items())
md = '## Migration lanes\n\n' + '\n'.join(['| lane | count |', '| --- | ---: |'] + [f'| {k} | {v} |' for k, v in rows])
display(Markdown(md))
```

```python
rows = sorted(archive['summary']['subdir_counts'].items())
md = '## Subdirectories\n\n' + '\n'.join(['| subdir | scripts |', '| --- | ---: |'] + [f'| `{k}` | {v} |' for k, v in rows])
display(Markdown(md))
```

```python
records = archive['records']
rows = []
for r in records[:60]:
    arts = len(r.get('sibling_artifacts', []))
    rows.append([f'[`{r["path"]}`](../../{r["path"]})', r['lane'], r['protected_reference_count'], arts, r['source']['line_count'], r['source']['sha256'][:12], r['source']['summary']])
md = '## Python inventory (first 60)\n\n' + '\n'.join(['| path | lane | refs | artifacts | lines | sha | summary |', '| --- | --- | ---: | ---: | ---: | --- | --- |'] + [f'| {a} | {b} | {c} | {d} | {e} | `{f}` | {g} |' for a,b,c,d,e,f,g in rows])
display(Markdown(md))
```

```python
protected = [r for r in records if r['protected_reference_count']]
rows = []
for r in protected[:30]:
    first = r['protected_references'][0] if r['protected_references'] else {}
    where = f'{first.get("path", "")}:{first.get("line", "")}' if first else '-'
    rows.append([f'[`{r["path"]}`](../../{r["path"]})', r['protected_reference_count'], where, r['lane']])
md = '## Protected validation references\n\n' + '\n'.join(['| script | refs | first ref | lane |', '| --- | ---: | --- | --- |'] + [f'| {a} | {b} | {c} | {d} |' for a,b,c,d in rows])
display(Markdown(md))
```

```python
assert archive['summary']['script_count'] == len(archive['records'])
assert all(r['source']['text'] for r in archive['records'])
assert archive['generated_at_utc']
print('peec_integration examples catalog synchronized')
```
