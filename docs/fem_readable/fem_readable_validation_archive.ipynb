# Readable FEM validation corpus archive

This notebook is a result-saved docs promotion of example Python scripts.
It intentionally pairs Markdown theory with executable inventory cells and an adjacent JSON debug record.

## Theory and placement
- This archive promotes the readable FEM validation scripts into docs while preserving their standalone teaching value.
- The corpus covers P1 tetrahedral and surface trace mechanics, Robin/Poisson assembly, RWG orientation, and force-load traces.
- Each script already writes a small summary JSON; the notebook gathers those summaries and keeps full source hashes next to the rendered result.

## Placement policy
KEEP for now per consolidation plan: readable validation scripts are a teaching corpus; move to validation_test only when each script gets an explicit regression gate.

## Synchronization policy
The durable archive is `fem_readable_validation_archive_results.json`. It stores full source, source SHA-256, existing result JSON payloads, runtime metadata, and the promotion lane for every script in this batch.

```python
from pathlib import Path
import json
from IPython.display import Markdown, display

archive_path = Path('fem_readable_validation_archive_results.json')
if not archive_path.exists():
    archive_path = Path('docs/fem_readable/fem_readable_validation_archive_results.json')
data = json.loads(archive_path.read_text(encoding='utf-8'))
counts = data['counts']
print(f"topic={data['topic']}")
print(f"python_scripts={counts['python_scripts']}")
print(f"attached_json_refs={counts['attached_json_refs']}")
print(f"topic_json_files={counts['topic_json_files']}")
print(f"promotion_lane={data['promotion_lane']}")
```

## Script inventory
The table below is generated from the synchronized JSON archive, not handwritten.

```python
rows = data['records']
header = '| # | script | lines | json refs | docstring |\n|---:|---|---:|---:|---|'
body = []
for i, rec in enumerate(rows, 1):
    doc = (rec.get('docstring_first_line') or '').replace('|', '\|')
    body.append(f"| {i} | `{rec['path']}` | {rec['line_count']} | {len(rec.get('attached_json', []))} | {doc} |")
display(Markdown('\n'.join([header] + body)))
```

## Integrity checks
These assertions make the notebook fail if the JSON/source pairing loses coverage.

```python
assert counts['python_scripts'] == len(data['records'])
assert all(len(rec['sha256']) == 64 for rec in data['records'])
assert all(rec.get('source') for rec in data['records'])
print('archive integrity: ok')
```

## Result payload index
Existing per-script JSON results remain in the archive so future debugging can compare old numbers with reruns.

```python
json_rows = data.get('topic_json_files', [])
if json_rows:
    header = '| # | json | sha256 | keys |\n|---:|---|---|---|'
    body = []
    for i, row in enumerate(json_rows, 1):
        payload = row.get('payload')
        keys = ', '.join(list(payload.keys())[:8]) if isinstance(payload, dict) else type(payload).__name__
        body.append(f"| {i} | `{row['path']}` | `{row['sha256'][:12]}` | {keys.replace('|', '\|')} |")
    display(Markdown('\n'.join([header] + body)))
else:
    print('no existing per-script JSON payloads found')
```
