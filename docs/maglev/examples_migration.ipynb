# Maglev Examples Migration

Maglev contains a curated public showcase, golden-test scripts, papers, and a large research validation corpus. The public demo assets route to docs/maglev/demos, paper assets to docs/maglev/papers, and validation/research CLN scripts to validation_test/maglev/research_cln.

This notebook is the result-saved migration ledger for `maglev`. It reads the adjacent JSON archive, verifies source coverage, and records the final target policy before `examples/maglev` is pruned.

```python
from pathlib import Path
import json
DATA = Path('examples_migration_results.json')
archive = json.loads(DATA.read_text(encoding='utf-8'))
summary = archive['summary']
summary
```

```python
from collections import Counter
files = archive['files']
assert archive['source_root'] == 'examples/maglev'
assert summary['files'] == len(files)
assert summary['python_files'] == sum(1 for f in files if f['suffix'] == '.py')
assert all(f['target_path'].startswith(('docs/', 'validation_test/')) for f in files)
Counter(f['target_path'].split('/')[0] + '/' + f['target_path'].split('/')[1] for f in files)
```

```python
from IPython.display import Markdown, display
rows = ['| source | target | bytes | sha256 |', '|---|---|---:|---|']
for f in files[:40]:
    rows.append(f"| `{f['path']}` | `{f['target_path']}` | {f['bytes']} | `{f['sha256'][:12]}` |")
if len(files) > 40:
    rows.append(f"| ... | {len(files)-40} more files archived in JSON |  |  |")
display(Markdown('\n'.join(rows)))
```
