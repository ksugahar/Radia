# PEEC Cleanup Routing

Result-saved routing notebook for the 63 `cleanup_review` scripts under `docs/peec_integration/demos`. This is the handoff table before moving files into `validation_test`, promoting reusable logic to `src`, or deleting one-off scratch after memory distillation.

```python
from pathlib import Path
import json
from IPython.display import Markdown, display
NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / 'cleanup_routing_results.json').exists():
    NOTEBOOK_DIR = Path('docs/peec_integration').resolve()
routing = json.loads((NOTEBOOK_DIR / 'cleanup_routing_results.json').read_text(encoding='utf-8'))
print(json.dumps(routing['summary'], indent=2, ensure_ascii=False))
```

```python
rows = sorted(routing['summary']['route_counts'].items())
md = '## Route Counts\n\n' + '\n'.join(['| target | count |', '| --- | ---: |'] + [f'| {k} | {v} |' for k, v in rows])
display(Markdown(md))
```

```python
for target in ['validation_test', 'src API', 'distill-delete']:
    rows = [r for r in routing['routes'] if r['target'] == target]
    table = ['| path | destination | reason |', '| --- | --- | --- |']
    for r in rows:
        table.append(f'| `{r["path"]}` | `{r["destination_hint"]}` | {r["reason"]} |')
    display(Markdown(f'## {target}\n\n' + '\n'.join(table)))
```

```python
protected = [r for r in routing['routes'] if r['protected_reference_count']]
rows = []
for r in protected[:25]:
    first = r['protected_references'][0] if r['protected_references'] else {}
    where = f'{first.get("path", "")}:{first.get("line", "")}' if first else '-'
    rows.append([r['path'], r['target'], r['protected_reference_count'], where])
md = '## Protected Reference Sample\n\n' + '\n'.join(['| path | target | refs | first ref |', '| --- | --- | ---: | --- |'] + [f'| `{a}` | {b} | {c} | {d} |' for a,b,c,d in rows])
display(Markdown(md))
```

```python
assert routing['summary']['cleanup_review_count'] == 63
assert sum(routing['summary']['route_counts'].values()) == 63
assert set(routing['summary']['route_counts']) == {'validation_test', 'src API', 'distill-delete'}
assert all(r['source_sha256'] for r in routing['routes'])
print('PEEC cleanup routing is synchronized')
```
