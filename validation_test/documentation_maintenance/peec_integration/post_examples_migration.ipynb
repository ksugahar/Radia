# PEEC Integration Post Examples Migration Catalog

Post-migration ledger for the remaining PEEC examples cleanup batch.

This notebook is result-saved. The adjacent JSON stores source hashes, selected source text, original paths, and post-migration destinations.


```python
import json
from pathlib import Path
from collections import Counter
import pandas as pd
data = json.loads(Path('post_examples_migration_results.json').read_text(encoding='utf-8'))
records = data['records']
assert len(records) == 193, len(records)
summary = data['summary']
summary
```

```python
rows = [{
    'original_path': r['original_path'],
    'destination': r['destination'],
    'lane': r['lane'],
    'action': r['action'],
    'suffix': r['suffix'],
    'size_bytes': r['size_bytes'],
} for r in records]
df = pd.DataFrame(rows)
display(df.groupby(['lane', 'action']).size().rename('count').reset_index())
display(df.head(20))
```

```python
dest_counts = Counter(r['destination'].split('/')[0] if r['destination'] else 'none' for r in records)
print('destination roots:', dict(dest_counts))
print('python scripts:', sum(1 for r in records if r['suffix'] == '.py'))
assert data['summary']['python_count'] == sum(1 for r in records if r['suffix'] == '.py')
```
