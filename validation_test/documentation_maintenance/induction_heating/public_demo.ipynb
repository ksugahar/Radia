# Induction Heating Public Demo

This notebook is the compact entry point after the induction-heating examples cleanup. It points readers to the saved demo showcase, validation lanes, and production APIs without depending on loose examples.

```python
from pathlib import Path
import json
from IPython.display import Markdown, display

data = json.loads(Path('public_demo_results.json').read_text(encoding='utf-8'))
data['closed_counts']
```

```python
def table(headers, rows):
    lines = ['| ' + ' | '.join(headers) + ' |', '|' + '|'.join(['---'] * len(headers)) + '|']
    lines += ['| ' + ' | '.join(str(c) for c in row) + ' |' for row in rows]
    return '\n'.join(lines)
rows = [[e['label'], e['path'], e['json'], e['sha256'][:12]] for e in data['entry_points']]
display(Markdown(table(['entry point', 'path', 'JSON/result', 'sha256'], rows)))
```
