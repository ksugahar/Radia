# Induction Heating Demo Showcase

This result-bearing notebook is the public docs replacement for the legacy induction-heating examples tier. It keeps the tutorial intent visible to readers while the executable implementation now lives in `src/radia`, validation fixtures, or this docs layer.

Policy points encoded here:

- public ESIM / WPT / RWG demonstration scripts are represented as a saved notebook plus synchronized JSON,
- legacy Cubit sample assets are routed to `validation_test/induction_heating/demoted_samples_legacy/`,
- no target path in this promotion ledger points back into `examples/`.

```python
from pathlib import Path
import json
from IPython.display import Markdown, display

RESULTS = Path('induction_heating_demo_showcase_results.json')
data = json.loads(RESULTS.read_text(encoding='utf-8'))
data['summary']
```

## Public Demonstration Layer

The five public scripts were tutorial surfaces around already-promoted APIs. The notebook records their source hashes and top-level functions so the teaching order remains visible without keeping a second live implementation under `examples`.

```python
def table(headers, rows):
    lines = ['| ' + ' | '.join(headers) + ' |', '|' + '|'.join(['---'] * len(headers)) + '|']
    lines += ['| ' + ' | '.join(str(c) for c in row) + ' |' for row in rows]
    return '\n'.join(lines)

rows = []
for rec in data['records']:
    funcs = ', '.join(rec.get('symbols', [])[:5])
    if len(rec.get('symbols', [])) > 5:
        funcs += ', ...'
    rows.append([rec['file_name'], rec['purpose'], rec['target'], funcs, rec['sha256'][:12]])
display(Markdown(table(['source', 'purpose', 'promotion target', 'top-level symbols', 'sha256'], rows)))
```

## Validation Fixture Layer

The Cubit journals and their Python generators are not final panel examples. They are preserved as legacy validation fixtures because they document Kelvin geometry construction, hole-SIBC setup, and historical closed-torus checks.

```python
rows = []
for rec in data['legacy_validation_records']:
    rows.append([rec['file_name'], rec['purpose'], rec['target'], rec['sha256'][:12]])
display(Markdown(table(['asset', 'why kept', 'new location', 'sha256'], rows)))
```

## Integrity Checks

The checks below make the promotion auditable. They deliberately verify counts, source preservation, and the absence of an examples destination before the original examples tree is removed.

```python
assert data['summary']['closed_tracked_files'] == 14
assert len(data['records']) == 5
assert len(data['legacy_validation_records']) == 7
assert len(data['readme_records']) == 2
assert data['checks']['has_public_demo_source_text']
assert data['checks']['has_legacy_source_text']
assert data['checks']['target_paths_are_non_examples']
print('promotion ledger checks: PASS')
```
