## Direction

mdx is a compute node. IH workflows should move from desktop PySide windows to JupyterHub/browser workbenches. The browser owns layout, fonts, zooming, and scroll behavior; solver code owns validated state, command generation, execution, and results.


## Layering

```text
radia.ih_design
  UI-neutral design state, method list, validation, spec -> calc command

radia.ih_runner
  run/cancel jobs, stream logs, parse JSON outputs, collect result paths

radia.ih_viewer
  webgui/results, tables, plots, snapshots

radia.ih_notebook
  ipywidgets/anywidget controls, design presets, sweeps, notebook/Voila layout
``` 


## Migration Rules

1. Do not add new IH behavior directly to `radia_ih.py`.
2. Put new user-facing IH inputs in a UI-neutral spec first.
3. Keep `calc_*.py` as the computation boundary.
4. Keep command-building tests independent of Qt.
5. Use the PySide panel only as a compatibility shell during migration.
6. Prefer browser widgets and 
gsolve.webgui` for the design workbench.
7. Treat `radia[gui]` as legacy; new design-tool work targets `radia[notebook]`.


## Packaging Target

```text
pip install radia
  core solvers, CLI, headless calc scripts

pip install radia[notebook]
  JupyterHub design workbench
  ipywidgets / anywidget / webgui widgets

pip install radia[legacy-gui]
  temporary compatibility extra for old PySide desktop panels
```
