# Stream-Function Public Demo Gallery

A result-saved public entry point for the stream-function demos. Validation and benchmark runners live under validation_test; this notebook keeps the human-facing demo path in docs while preserving the runnable examples as companions.

````python
from pathlib import Path
import json
from IPython.display import Markdown, display
NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / 'demo_gallery_results.json').exists():
    NOTEBOOK_DIR = Path('docs/stream_function').resolve()
archive = json.loads((NOTEBOOK_DIR / 'demo_gallery_results.json').read_text(encoding='utf-8'))
print(json.dumps(archive['summary'], indent=2, ensure_ascii=False))
````

````python
from collections import Counter
records = archive['records']
rows = sorted(archive['summary']['demo_group_counts'].items())
md = '## Demo groups\n\n' + '\n'.join(['| group | count |', '| --- | --- |'] + [f'| {k} | {v} |' for k, v in rows])
display(Markdown(md))
````

````python
rows = []
commands = {
    'demo_sf_to_peec_gz.py': 'python docs/stream_function/demo_sf_to_peec_gz.py --with-peec',
    'demo_sf_to_peec_gx.py': 'python docs/stream_function/demo_sf_to_peec_gx.py --chain-method kuijpers',
    'demo_planar_uniform_fem_psi.py': 'python docs/stream_function/demo_planar_uniform_fem_psi.py --regularize h1 --compensated-iter 100 --compensated-step 0.05',
    'demo_sphere_fe_direct.py': 'python docs/stream_function/demo_sphere_fe_direct.py --target z2 --distort --plot',
    'demo_pareto_tikhonov_aca.py': 'python docs/stream_function/demo_pareto_tikhonov_aca.py --front both',
    'demo_regcoil_fusion.py': 'python docs/stream_function/demo_regcoil_fusion.py',
    'view_sf_coil_gx_gmsh.py': 'python docs/stream_function/view_sf_coil_gx_gmsh.py --mode contours',
}
for r in archive['selected_gallery']:
    arts = ', '.join(f'[{Path(a.get("current_path", a["path"])).name}]({a.get("current_path", a["path"]).split("docs/stream_function/")[-1]})' for a in r.get('sibling_artifacts', [])[:3]) or '-'
    rows.append([f'[`{r["name"]}`]({r.get("current_path", r["path"]).split("docs/stream_function/")[-1]})', r['demo_group'], f'`{commands.get(r["name"], "python docs/stream_function/" + r["name"])}`', arts, r['source']['summary']])
md = '## Public learning path\n\n' + '\n'.join(['| demo | group | command | saved results | source summary |', '| --- | --- | --- | --- | --- |'] + [f'| {a} | {b} | {c} | {d} | {e} |' for a,b,c,d,e in rows])
display(Markdown(md))
````

````python
artifact_rows = []
for r in records:
    for a in r.get('sibling_artifacts', []):
        if a['path'].endswith(('.png', '.pdf', '.json')):
            artifact_rows.append([f'[`{r["name"]}`]({r.get("current_path", r["path"]).split("docs/stream_function/")[-1]})', f'[{Path(a.get("current_path", a["path"])).name}]({a.get("current_path", a["path"]).split("docs/stream_function/")[-1]})', a['bytes'], a['sha256'][:12]])
artifact_rows = artifact_rows[:18]
md = '## Saved result/media artifacts (sample)\n\n' + '\n'.join(['| script | artifact | bytes | sha256 |', '| --- | --- | ---: | --- |'] + [f'| {a} | {b} | {c} | `{d}` |' for a,b,c,d in artifact_rows])
display(Markdown(md))
````

````python
for r in archive['selected_gallery'][:3]:
    excerpt = r['source']['text'][:1200].replace('```', '` ` `')
    display(Markdown(f'## Source excerpt: `{r["path"]}`\n\n```python\n{excerpt}\n```'))
````

````python
assert archive['summary']['script_count'] == len(archive['records'])
assert all(r['source']['text'] for r in archive['records'])
assert archive['policy']['validation_split']
print('stream_function public demo gallery synchronized')
````
