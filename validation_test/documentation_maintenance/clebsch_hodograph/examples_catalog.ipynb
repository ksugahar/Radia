# Clebsch Hodograph Examples Catalog

A result-saved public demo ledger for the Clebsch/hodograph research line. The notebook ties the existing theory notes to the runnable example sources, saved JSON/figure artifacts, and the protected validation references that must be handled before pruning examples.

````python
from pathlib import Path
import json
from IPython.display import Markdown, display
NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / 'examples_catalog_results.json').exists():
    NOTEBOOK_DIR = Path('docs/clebsch_hodograph').resolve()
archive = json.loads((NOTEBOOK_DIR / 'examples_catalog_results.json').read_text(encoding='utf-8'))
print(json.dumps(archive['summary'], indent=2, ensure_ascii=False))
````

````python
rows = sorted(archive['summary']['lane_counts'].items())
md = '## Hodograph demo lanes\n\n' + '\n'.join(['| lane | count |', '| --- | ---: |'] + [f'| {k} | {v} |' for k, v in rows])
display(Markdown(md))
````

````python
records = archive['records']
featured_names = [
    'HODOGRAPH_BACKBONE.md',
    'hodograph_kelvin_2d.py',
    'hodograph_kelvin_axisym.py',
    'clebsch_kelvin_3d.py',
    'chaplygin_hodograph_2d.py',
    'a_method_clebsch_2d.py',
    'accel_pole_design.py',
    'endpack_cobake_loft.py',
]
rows = []
for name in featured_names[1:]:
    r = next((x for x in records if x['name'] == name), None)
    if not r:
        continue
    arts = ', '.join(f'[{Path(a["path"]).name}](../../{a["path"]})' for a in r.get('sibling_artifacts', [])[:3]) or '-'
    rows.append([f'[`{r["name"]}`](../../{r["path"]})', r['lane'], arts, r['source']['summary']])
md = '## Public demo spine\n\nStart from [`HODOGRAPH_BACKBONE.md`](HODOGRAPH_BACKBONE.md), then inspect these runnable companions.\n\n' + '\n'.join(['| demo | lane | saved results | source summary |', '| --- | --- | --- | --- |'] + [f'| {a} | {b} | {c} | {d} |' for a,b,c,d in rows])
display(Markdown(md))
````

````python
artifact_rows = []
for r in records:
    for a in r.get('sibling_artifacts', []):
        artifact_rows.append([f'[`{r["name"]}`](../../{r["path"]})', f'[{Path(a["path"]).name}](../../{a["path"]})', a['bytes'], a['sha256'][:12]])
artifact_rows = artifact_rows[:30]
md = '## Saved result/media artifacts (sample)\n\n' + '\n'.join(['| script | artifact | bytes | sha256 |', '| --- | --- | ---: | --- |'] + [f'| {a} | {b} | {c} | `{d}` |' for a,b,c,d in artifact_rows])
display(Markdown(md))
````

````python
protected = [r for r in records if r['protected_reference_count']]
rows = []
for r in protected[:20]:
    first = r['protected_references'][0] if r['protected_references'] else {}
    where = f'{first.get("path", "")}:{first.get("line", "")}' if first else '-'
    rows.append([f'[`{r["name"]}`](../../{r["path"]})', r['protected_reference_count'], where, r['lane']])
md = '## Protected validation references\n\n' + '\n'.join(['| script | refs | first ref | lane |', '| --- | ---: | --- | --- |'] + [f'| {a} | {b} | {c} | {d} |' for a,b,c,d in rows])
display(Markdown(md))
````

````python
for name in ['hodograph_kelvin_2d.py', 'chaplygin_hodograph_2d.py', 'accel_pole_design.py']:
    r = next(x for x in records if x['name'] == name)
    excerpt = r['source']['text'][:1200].replace('```', '` ` `')
    display(Markdown(f'## Source excerpt: `{r["path"]}`\n\n```python\n{excerpt}\n```'))
````

````python
current_paths = {r['path'] for r in archive['records']}
assert archive['summary']['script_count'] == len(current_paths)
assert all(r['source']['text'] for r in archive['records'])
assert archive['generated_at_utc']
print('clebsch_hodograph public demo catalog synchronized')
````
